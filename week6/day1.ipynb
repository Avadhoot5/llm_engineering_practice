```python
# imports

import os
from dotenv import load_dotenv
from litellm import completion
from huggingface_hub import login
from datasets import load_dataset
import matplotlib.pyplot as plt
from tqdm.notebook import tqdm
import numpy as np
import random
from week6.pricer.items import Item
from week6.pricer.parser import parse
```

```python
# Log in to HuggingFace - if you get a "Note" about Environment variable being set, ignore it

hf_token = os.environ['HF_TOKEN']
login(hf_token, add_to_git_credential=True)
```

```python
load_dotenv(override=True)

OLLAMA_API_KEY = os.getenv('OLLAMA_API_KEY')

ollama_url = "https://ollama.com"

MODEL_OLLAMA = "ollama/gpt-oss:120b-cloud"
```

```python
messages = [{'role': 'user', 'content': 'what is  2+ 2'}]

response = completion(model=MODEL_OLLAMA, api_base=ollama_url, api_key=OLLAMA_API_KEY, messages= messages)

print(response.choices[0].message.content)
```
